### **Summary of Findings**

This exploratory data analysis of the *Mental Health in Tech* survey offers a comprehensive view of how mental health is perceived, discussed, and supported within the technology sector. Overall, the EDA demonstrates a clear pattern: mental health concerns are widespread in the tech industry, but support is unevenly distributed across demographics, company size, and geographic regions. Stigma, lack of resources, and inconsistent managerial support continue to hinder open discussion and access to care. The findings emphasize the need for proactive, well-structured mental health programs that encourage open communication, protect employee confidentiality, and treat mental health with the same seriousness as physical health.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Make dataset analysis ready

# ---------------- STEP 1: Clean text formatting ----------------
df = df.applymap(lambda x: x.strip().lower() if isinstance(x, str) else x)


# ---------------- STEP 2: Fix age column ----------------
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Remove unrealistic ages (like below 15 or above 80)
df = df[(df['Age'] >= 15) & (df['Age'] <= 80)]


# ---------------- STEP 3: Handle missing categorical values ----------------
categorical_cols = df.select_dtypes(include='object').columns
df[categorical_cols] = df[categorical_cols].fillna('unknown')


# ---------------- STEP 4: Handle missing numeric values ----------------
numeric_cols = df.select_dtypes(include=['int64','float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())


# ---------------- STEP 5: Encode categorical variables ----------------
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])


# ---------------- STEP 6: Verify clean dataset ----------------
print("Dataset is now cleaned and analysis ready!")
print("\nShape:", df.shape)
print("\nMissing values after cleaning:\n")
print(df.isnull().sum())
````

### What all manipulations have you done and insights you found?


To prepare the dataset for analysis, several preprocessing steps were performed:

* **Text Cleaning:** All text values were converted to lowercase and extra spaces were removed to standardize the data.
* **Age Correction:** The age column contained unrealistic values (very low and very high), so rows outside the range of **15–80** were removed.
* **Handling Missing Values:**

  * Missing categorical values were replaced with "unknown"
  * Missing numerical values were filled with the **median**
* **Duplicate Check:** The dataset was checked for duplicate records.
* **Encoding Categorical Data:** Since machine learning models work with numbers, categorical text fields were transformed using **Label Encoding**.
